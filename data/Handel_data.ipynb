```python
TRAIN_ROOT = r"D:\archive\BraTS2020_TrainingData\MICCAI_BraTS2020_TrainingData"
LABELS_CSV = r"D:\archive\BraTS2020_TrainingData\labels_with_survival.csv"
OUT_ROOT   = r"D:\archive\BraTS2020_2D_dataset"   # هيطلع فيه صور 2D
```

```python
import os, glob, pandas as pd

labels = pd.read_csv(LABELS_CSV)
labels['label'] = labels['label'].str.strip().str.upper()  # توحيد الشكل
subjects = {row['subject']: row['label'] for _, row in labels.iterrows()}

# نجيب كل مجلدات الحالات الموجودة فعليًا
case_dirs = [d for d in os.listdir(TRAIN_ROOT) if d.startswith("BraTS20_")]
present = [c for c in case_dirs if c in subjects]

missing_in_labels = [c for c in case_dirs if c not in subjects]
print(f"✅ حالات موجودة ولها label: {len(present)}")
print(f"⚠️ حالات بلا label (لو فيه): {len(missing_in_labels)} -> {missing_in_labels[:5]}")
```

```python
import nibabel as nib
import numpy as np
from skimage.transform import resize
import imageio.v3 as iio

def find_flair_path(case_dir):
    # يقبل .nii أو .nii.gz
    p1 = glob.glob(os.path.join(case_dir, "*_flair.nii"))
    p2 = glob.glob(os.path.join(case_dir, "*_flair.nii.gz"))
    return (p1 + p2)[0] if (p1 + p2) else None

def normalize_slice(x):
    # قص الأطراف للحد من القيم الشاذة ثم min-max
    lo, hi = np.percentile(x, 0.5), np.percentile(x, 99.5)
    x = np.clip(x, lo, hi)
    x = x - x.min()
    if x.max() > 0: x = x / x.max()
    return x

def volume_to_slices(vol, k=7):
    z = vol.shape[2]
    mid = z // 2
    half = k // 2
    idxs = list(range(max(0, mid-half), min(z, mid+half+1)))
    slices = [normalize_slice(vol[:, :, i]) for i in idxs]
    return slices

def save_slices(slices, out_dir, base_name):
    os.makedirs(out_dir, exist_ok=True)
    for j, sl in enumerate(slices):
        sl_res = resize(sl, (224, 224), anti_aliasing=True, preserve_range=True)
        # إلى 8-بت PNG
        img8 = (sl_res * 255).astype(np.uint8)
        iio.imwrite(os.path.join(out_dir, f"{base_name}_z{j:02d}.png"), img8)
```

```python
from tqdm import tqdm

counts = {"HGG": 0, "LGG": 0}
saved  = 0
skipped = []

for case in tqdm(present):
    case_dir = os.path.join(TRAIN_ROOT, case)
    flair = find_flair_path(case_dir)
    if not flair:
        skipped.append((case, "no_flair"))
        continue

    try:
        img = nib.load(flair)
        vol = img.get_fdata().astype(np.float32)

        # لو الفوليوم كله صفران نتخطاه
        if np.all(vol == 0) or np.nanmax(vol) == 0:
            skipped.append((case, "all_zero"))
            continue

        slices = volume_to_slices(vol, k=7)
        label  = subjects[case]  # HGG / LGG
        out_dir = os.path.join(OUT_ROOT, label)
        save_slices(slices, out_dir, base_name=case)
        counts[label] += 1
        saved += len(slices)
    except Exception as e:
        skipped.append((case, str(e)))

print("✅ حالات تم استخراج شرائح لها:", counts)
print(f"✅ إجمالي صور 2D المحفوظة: {saved}")
print("⚠️ حالات متخطّاة:", len(skipped), "أول 5:", skipped[:5])
print(f"📁 خرج الصور: {OUT_ROOT}\\HGG و {OUT_ROOT}\\LGG")
```

```python
import os
import shutil
import random

# المسارات
SRC_DIR = r"D:\archive\BraTS2020_2D_dataset"
DEST_DIR = r"D:\archive\BraTS2020_2D_split"

# النسب
train_ratio = 0.7
val_ratio   = 0.2
test_ratio  = 0.1

# إنشاء المجلدات الجديدة
for split in ['train', 'val', 'test']:
    for cls in ['HGG', 'LGG']:
        os.makedirs(os.path.join(DEST_DIR, split, cls), exist_ok=True)

# تقسيم كل فئة على حدة
for cls in ['HGG', 'LGG']:
    src_path = os.path.join(SRC_DIR, cls)
    imgs = os.listdir(src_path)
    random.shuffle(imgs)

    n_total = len(imgs)
    n_train = int(train_ratio * n_total)
    n_val   = int(val_ratio * n_total)
    
    train_imgs = imgs[:n_train]
    val_imgs   = imgs[n_train:n_train + n_val]
    test_imgs  = imgs[n_train + n_val:]

    # نسخ الملفات
    for img in train_imgs:
        shutil.copy(os.path.join(src_path, img), os.path.join(DEST_DIR, 'train', cls, img))
    for img in val_imgs:
        shutil.copy(os.path.join(src_path, img), os.path.join(DEST_DIR, 'val', cls, img))
    for img in test_imgs:
        shutil.copy(os.path.join(src_path, img), os.path.join(DEST_DIR, 'test', cls, img))

print("✅ تم تقسيم البيانات بنجاح!")
```

```python
# ===============================
# 📦 Data Loader for BraTS2020 Segmentation (Fixed MONAI v1+ Compatible)
# ===============================

import os
import glob
import torch
import nibabel as nib
import numpy as np
from torch.utils.data import Dataset, DataLoader
from config_seg import DATA_DIR, PATCH_SIZE, BATCH_SIZE, MODALITY


# ===============================
# 🧩 Custom Dataset Class
# ===============================
class BraTSDataset(Dataset):
    def __init__(self, image_paths, label_paths, transforms=None):
        self.image_paths = image_paths
        self.label_paths = label_paths
        self.transforms = transforms

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        label_path = self.label_paths[idx]

        # تحميل الصورة والماسك
        image = nib.load(image_path).get_fdata().astype(np.float32)
        label = nib.load(label_path).get_fdata().astype(np.float32)

        # تطبيع الصورة
        image = np.clip(image, 0, 2000)
        image = image / 2000.0

        # إضافة بعد القناة
        image = np.expand_dims(image, axis=0)
        label = np.expand_dims(label, axis=0)

        # تطبيق التحويلات البسيطة (لو موجودة)
        if self.transforms:
            image, label = self.transforms(image, label)

        return torch.tensor(image), torch.tensor(label)


# ===============================
# ⚙️ Function to create dataloaders
# ===============================
def get_datasets():
    images = sorted(glob.glob(os.path.join(DATA_DIR, f"*/BraTS20_*_{MODALITY}.nii")))
    labels = sorted(glob.glob(os.path.join(DATA_DIR, "*/BraTS20_*_seg.nii")))

    n = len(images)
    split = int(n * 0.8)

    train_images, val_images = images[:split], images[split:]
    train_labels, val_labels = labels[:split], labels[split:]

    print(f"✅ Found {len(train_images)} training cases and {len(val_images)} validation cases.")

    # 🧠 Transform بسيط (قص عشوائي)
    def simple_transform(image, label):
        x, y, z = image.shape[1:]
        start_x = np.random.randint(0, x - PATCH_SIZE[0])
        start_y = np.random.randint(0, y - PATCH_SIZE[1])
        start_z = np.random.randint(0, z - PATCH_SIZE[2])

        image_crop = image[:, start_x:start_x + PATCH_SIZE[0],
                           start_y:start_y + PATCH_SIZE[1],
                           start_z:start_z + PATCH_SIZE[2]]
        label_crop = label[:, start_x:start_x + PATCH_SIZE[0],
                           start_y:start_y + PATCH_SIZE[1],
                           start_z:start_z + PATCH_SIZE[2]]

        return image_crop, label_crop

    # إنشاء Datasets
    train_ds = BraTSDataset(train_images, train_labels, transforms=simple_transform)
    val_ds = BraTSDataset(val_images, val_labels, transforms=None)

    # إنشاء DataLoaders
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=1, shuffle=False, num_workers=0)

    return train_loader, val_loader


# ===============================
# 🔍 اختبار سريع
# ===============================
if __name__ == "__main__":
    train_loader, val_loader = get_datasets()
    img, mask = next(iter(train_loader))
    print("🧩 Image shape:", img.shape)
    print("🧩 Mask shape:", mask.shape)
```
